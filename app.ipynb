```python
import os
import shutil
import pandas as pd
import kagglehub

DATA_DIR = "data"
CSV_PATH = os.path.join(DATA_DIR, "creditcard.csv")

def is_valid_csv(path):
    """Return True if file exists and is a readable, non-empty CSV."""
    if not os.path.isfile(path):
        return False
    if os.path.getsize(path) < 1024:  # sanity: too small to be real
        return False
    try:
        # Peek at the file to ensure it's parseable
        pd.read_csv(path, nrows=5)
        return True
    except Exception:
        return False

def loader():
    print(f"{CSV_PATH} missing or corrupted, downloading...")
    os.makedirs(DATA_DIR, exist_ok=True)

    dataset_path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")
    src = os.path.join(dataset_path, "creditcard.csv")

    # If cache gives a zip instead of a folder, handle it
    if src.endswith(".zip") or not os.path.isfile(src):
        # find the csv inside dataset_path (handle nested dirs)
        for root, _, files in os.walk(dataset_path):
            if "creditcard.csv" in files:
                src = os.path.join(root, "creditcard.csv")
                break

    shutil.copy(src, CSV_PATH)
    print(f"Saved to {CSV_PATH}")

if is_valid_csv(CSV_PATH):
    print(f"Found valid file at {CSV_PATH}, skipping download.")
else:
  loader()

df = pd.read_csv(CSV_PATH)
```
